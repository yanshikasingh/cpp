# UNIT 2 — Data Analytics, Pre-processing, PCA, NumPy, Pandas & SciPy

## 1. Multiple Linear Regression — Statsmodels

```python
import pandas as pd
import statsmodels.api as sm

# Load dataset
data = pd.read_csv('data.csv')

# Dependent variable
Y = data['Y']

# Independent variables
X = data[['X1', 'X2', 'X3']]

# Add constant/intercept term
X = sm.add_constant(X)

# Fit multiple linear regression model
model = sm.OLS(Y, X).fit()

# Print regression summary
print(model.summary())
```

### 2. Data Pre-processing — Load Dataset and Inspect

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import seaborn as sns
import matplotlib.pyplot as plt

# Change this path to your CSV file
df = pd.read_csv('diabetes.csv')

# View first 5 rows
df.head()

# Inspect structure
df.info()

# Check missing values
print(df.isnull().sum())
```

## 3. Replacing Missing Values

```python
# Mean imputation for the Insulin column
df['Insulin'] = df['Insulin'].fillna(df['Insulin'].mean())

# Other common forms:
# df['column'] = df['column'].fillna(value)
# df['column'] = df['column'].fillna(df['column'].mean())
# df['column'] = df['column'].fillna(df['column'].median())
# df = df.ffill()   # forward fill
# df = df.bfill()   # backward fill

print(df.isnull().sum())
```

## 4. Save Updated CSV and Excel Files

```python
df.to_csv('updated_dataset.csv', index=False)
df.to_excel('updated_dataset.xlsx', index=False)

print("Files saved successfully.")
```

## 5. Checking and Removing Duplicates

```python
# Check duplicate rows
duplicates = df.duplicated()
print(duplicates)

# Remove duplicates based on one column
df_no_duplicates_columns = df.drop_duplicates(subset=['Insulin'])
print(df_no_duplicates_columns)

# Remove duplicate rows, keeping the first occurrence
df_cleaned = df.drop_duplicates(keep='first')

# Save cleaned data
df_cleaned.to_csv('updated_dataset.csv', index=False)
```

## 6. Statistical Summary and Visualizing Outliers

```python
# Statistical summary
print(df.describe())

# Boxplots for all columns
fig, axs = plt.subplots(len(df.columns), 1, figsize=(7, 18), dpi=95)

for i, col in enumerate(df.columns):
    axs[i].boxplot(df[col].dropna(), vert=False)
    axs[i].set_ylabel(col)

plt.tight_layout()
plt.show()
```

## 7. Remove Outliers Using IQR

```python
# IQR method for the Insulin column
q1, q3 = np.percentile(df['Insulin'].dropna(), [25, 75])

iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

clean_df = df[(df['Insulin'] >= lower) & (df['Insulin'] <= upper)]

print(clean_df.head())
```

## 8. Data Transformation — Rename, Create, Filter, Type Conversion, Grouping

```python
# Rename a column
df.rename(columns={'old_column_name': 'new_column_name'}, inplace=True)

# Create a new column
df['total_price'] = df['quantity'] * df['unit_price']

# Filter rows
df = df[df['quantity'] > 10]

# Change data type
df['quantity'] = df['quantity'].astype(int)

# Group and aggregate
summary_df = df.groupby('category')['total_price'].sum().reset_index()

# Save transformed data
df.to_csv('updated_dataset.csv', index=False)

print(summary_df)
```

# 13. Principal Component Analysis (PCA) — Iris Dataset

This follows the PCA workflow shown in the Unit-2 material: import libraries → load Iris → standardize → correlation heatmap → apply PCA → inspect reduced components → correlation heatmap after PCA.

```python
# Step 1: Import necessary libraries
from sklearn import datasets
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Step 2: Load the Iris dataset
iris = datasets.load_iris()

# Convert the dataset into a pandas DataFrame
df_iris = pd.DataFrame(
    iris['data'],
    columns=iris['feature_names']
)

# Display first 5 rows
df_iris.head()
```

```python
# Step 3: Standardize the features
scalar = StandardScaler()

scaled_data = pd.DataFrame(
    scalar.fit_transform(df_iris),
    columns=df_iris.columns
)

scaled_data.head()
```

```python
# Step 3 (optional): Correlation before PCA
sns.heatmap(scaled_data.corr(), annot=True)
plt.show()
```

```python
# Step 4: Apply PCA
pca = PCA(n_components=3)

pca.fit(scaled_data)

data_pca = pca.transform(scaled_data)

data_pca = pd.DataFrame(
    data_pca,
    columns=['PC1', 'PC2', 'PC3']
)

data_pca.head()
```

```python
# Step 5: Correlation after PCA
sns.heatmap(data_pca.corr(), annot=True)
plt.show()
```

## 14. PCA Visualization of Iris Dataset — Unit 3 Example

```python
from sklearn import datasets, decomposition
import seaborn as sns
import matplotlib.pyplot as plt

# Load Iris dataset
iris = datasets.load_iris()

X = iris.data
y = iris.target

# Reduce dimensions to 2
pca = decomposition.PCA(n_components=2)
X_transformed = pca.fit_transform(X)

# Scatter plot of reduced dimensions
sns.scatterplot(
    x=X_transformed[:, 0],
    y=X_transformed[:, 1],
    hue=y
)

plt.show()
```

## 34. 2×2 Subplots Using `subplots()`

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 400)

y1 = np.sin(x)
y2 = np.sin(x**2)
y3 = y1**2
y4 = y2**2

fig, ax = plt.subplots(nrows=2, ncols=2)

ax[0, 0].plot(x, y1, c='red')
ax[0, 1].plot(x, y2, c='red')
ax[1, 0].plot(x, y3, c='blue')
ax[1, 1].plot(x, y4, c='blue')

ax[0, 0].set_title('Simple plot with sin(x)')
ax[0, 1].set_title('Simple plot with sin(x**2)')
ax[1, 0].set_title('Simple plot with sin(x)**2')
ax[1, 1].set_title('Simple plot with sin(x**2)**2')

fig.suptitle('Stacked subplots in two directions')
plt.show()
```

## 35. Subplots with Shared X-Axis

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 400)

y1 = np.sin(x)
y2 = np.sin(x**2)

fig, (ax1, ax2) = plt.subplots(2, sharex=True)

ax1.plot(x, y1, c='red')
ax2.plot(x, y2, c='red')

ax1.set_ylabel('Simple plot with sin(x)')
ax2.set_ylabel('Simple plot with sin(x**2)')

fig.suptitle('Subplots with shared axis')
plt.show()
```
